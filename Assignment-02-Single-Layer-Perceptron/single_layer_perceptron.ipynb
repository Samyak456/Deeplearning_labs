```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# OR Gate Dataset

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([
    0,
    1,
    1,
    1
])

print("Input Data:")
print(X)

print("\nTarget Output:")
print(y)
```

```python
def step_function(x):
    if x >= 0:
        return 1
    else:
        return 0
```

```python
# Initialize weights and bias

weights = np.zeros(2)
bias = 0

learning_rate = 0.1
epochs = 10

print("Initial Weights:", weights)
print("Initial Bias:", bias)

for epoch in range(epochs):

    for i in range(len(X)):

        linear_output = np.dot(X[i], weights) + bias

        prediction = step_function(linear_output)

        error = y[i] - prediction

        weights += learning_rate * error * X[i]

        bias += learning_rate * error

    print(f"Epoch {epoch+1}")
    print("Weights:", weights)
    print("Bias:", bias)
```

```python
print("Testing the Perceptron\n")

for i in range(len(X)):

    linear_output = np.dot(X[i], weights) + bias

    prediction = step_function(linear_output)

    print(f"Input: {X[i]} --> Predicted: {prediction} | Actual: {y[i]}")
```

```python
test_cases = [
    ([0.2,0.2], -0.1),
    ([1,1], -0.5),
    ([0.5,0.5], 0),
    ([-1,-1], 0.5)
]

for w,b in test_cases:

    print("="*40)
    print("Weights:", w)
    print("Bias:", b)

    for sample in X:

        output = np.dot(sample,w)+b

        prediction = step_function(output)

        print(sample,"->",prediction)
```

```python
def step(x):
    return 1 if x>=0 else 0

def sigmoid(x):
    return 1/(1+np.exp(-x))

def relu(x):
    return max(0,x)

print("Activation Function Comparison\n")

for sample in X:

    z = np.dot(sample,weights)+bias

    print("Input:",sample)

    print("Linear Output:",round(z,3))

    print("Step:",step(z))

    print("Sigmoid:",round(sigmoid(z),3))

    print("ReLU:",round(relu(z),3))

    print()
```
